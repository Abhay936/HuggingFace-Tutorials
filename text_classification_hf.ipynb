```python
import transformers
```

```python
from transformers import pipeline
```

```python
spam_classification_pipeline = pipeline("text-classification",model="FaizAhmadDev/sms-spam-detection-distilbert")
```

```python
texts = [
    "Congratulations! You have won a $1,000 prize. Click here to claim now!",
    "Hey, are we still meeting at 5 PM today?",
    "URGENT! You have been selected for a cash reward. Claim it now!",
    "Please send me the project report when you get time.",
    "Get 50% OFF today! Limited time offer. Click this link now!"
]

results = spam_classification_pipeline(texts)

labels = {
    "LABEL_0": "Ham",
    "LABEL_1": "Spam"
}

for result in results:
    label = labels[result["label"]]
    score = round(result["score"], 4)
    print(f"Label : {label}, Confidence : {score}")
```
