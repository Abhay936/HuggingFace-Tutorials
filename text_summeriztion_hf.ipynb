```python
import transformers
```

```python
from transformers import AutoModelForSeq2SeqLM, AutoTokenizer
```

```python
model_name = 'facebook/bart-large-cnn'
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)
```

```python
text = "Machine learning is a branch of artificial intelligence that allows computers to learn patterns from data without being explicitly programmed for every task. It is widely used in applications such as recommendation systems, spam detection, fraud detection, image recognition, and predictive analytics. Machine learning models generally improve their performance as they are trained on more relevant and high-quality data."
```

```python
len(text)
```

```python
inputs = tokenizer.encode(text, return_tensors='pt', max_length=512, truncation=True)
```

```python
summary_ids = model.generate(inputs, max_length=50, min_length=25, length_penalty=2.0, num_beams=4, early_stopping=True)
```

```python
summary = tokenizer.decode(summary_ids[0], skip_special_tokens=True)
```

```python
print(summary)
```
